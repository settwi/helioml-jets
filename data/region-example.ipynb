# Example: importing the region cutout data, and plotting some bounding boxes
- Assumes you have `jet_cutouts.asdf` from the Google Drive link in the readme file,
or that you have generated your own.
- Assumes you have your Python environment set up properly

```python
import asdf
import matplotlib.pyplot as plt
import shapely
```

```python
af = asdf.open("jet_cutouts.asdf")
# Print the keys/file layout
af.info()
```

```python
# Find the entry with the most images and bounding boxes
max_ = 0
best = None
for k in af.keys():
    if not k.startswith("sjh"):
        continue

    if (cur_size := len(af[k]["raw_cutouts"])) > max_:
        best = k
        max_ = cur_size
```

```python
# Each entry has the "raw" cutout, the corners of the bounding box region,
# and the cutouts scaled by the default SunPy color maps
af[best].keys()
```

```python
# Make plots of the cutouts/boxes
%matplotlib inline
i = 0
for cutout, corners in zip(af[best]["scaled_cutouts"], af[best]["region_corners"]):
    # make a shapely object for manipulation later
    rect = shapely.Polygon(corners)
    fig, ax = plt.subplots()

    ax.imshow(cutout, cmap="Blues")
    ax.plot(*rect.exterior.xy, color="red")
    ax.set(title=f"image {i}")

    ax.set(xlim=(0, cutout.shape[0]), ylim=(0, cutout.shape[1]))

    plt.show()
    plt.close(fig)
    i += 1
```

```python
# Use the "raw" cutouts to see the difference
%matplotlib inline
i = 0
for cutout, corners in zip(af[best]["raw_cutouts"], af[best]["region_corners"]):
    # make a shapely object for manipulation later
    rect = shapely.Polygon(corners)
    fig, ax = plt.subplots()

    ax.imshow(cutout, cmap="Blues")
    ax.plot(*rect.exterior.xy, color="red")
    ax.set(title=f"image {i}")

    ax.set(xlim=(0, cutout.shape[0]), ylim=(0, cutout.shape[1]))

    plt.show()
    plt.close(fig)
    i += 1
```
